# Introducción
---

El consumo de café es un producto que puede generar placer, energía, incluso ayudar al estado de ánimo, sin embargo este también puede afectar la salud, incluyendo el riesgo de enfermedades cardiovasculares, metabólicas y neurológicas e incluso, en ciertos casos, generar dependencia.

El problema general que aborda este proyecto es entender la relación entre el consumo de café y diferentes factores de salud sobretodo con el nivel de estrés, lo cual es relevante ya que ayuda a prevenir enfermedades, creando decisiones informadas dentro de la industria de nutrición.

El alcance de esta primera entrega se centra en realizar un análisis exploratorio descriptivo (EDA) para identificar patrones generales y entender los datos con los que se estarán trabajando, también se realizará una organización de los datos donde se hará la limpieza de los mismos para poder trabajar con datos limpios.

Se trabajará con el dataset *GlobalCoffeeHealth* desde Kaggle el cual contiene 10.000 registros sintéticos que reflejan patrones reales de consumo de café, hábitos de sueño, niveles de estrés y resultados de salud en 20 países como Alemania, Brasil, México, entre otros. 

El dataset fue obtenido del siguiente link: https://www.kaggle.com/datasets/uom190346a/global-coffee-health-dataset/data

# Preprocesamiento de Datos
---

El preprocesamiento se realiza por modulos mediante dos funciones principales:  
`preprocessing_train()`, para entrenamiento y `preprocessing_eval()`, para validación y prueba.  
Ambas transforman los datos de forma consistente, solo la primera ajusta los objetos (**DictVectorizer**, **StandardScaler**, **SMOTE**) y define las features finales.

## Flujo del Prepocesamiento
**Entrenamiento (`preprocessing_train`)**
1. **Limpieza de columnas:**  
   Se eliminan las variables: *Health_Issues*, *Caffeine_mg*, *Sleep_Hours*, *Sleep_Quality*
2. **Filtrado de datos:**  
   Se excluyen registros con *Gender = "Other"*.
3. **Codificación categórica manual:**  
   - *Country* se transforma a *Continent* mediante un mapeo (América, Europa, Asia, Oceanía).  
   - *Gender* se convierte a binaria (*0 = Male*, *1 = Female*).  
   - *Stress_Level* se codifica ordinalmente (*Low=0*, *Medium=1*, *High=2*).
4. **Codificación categórica general:**  
   Se aplica un `DictVectorizer` para convertir todas las categorías en variables binarias.
5. **Reducción de colinealidad:**  
   - Se elimina un dummy por categoría (para evitar colinealidad perfecta).  
   - Se eliminan variables con alta correlación (|r| > 0.9) usando Pearson.
6. **Selección de características:**  
   Se calculó la **información mutua** (*mutual_info_classif*) para identificar las variables con mayor dependencia con el target.  
   Se conservaron las **20 más informativas**, de las cuales *15* pasaron los filtros de correlación y redundancia.
7. **Estandarización:**  
   Se aplica `RobustScaler` a las variables seleccionadas.
8. **Balanceo de clases:**  
   Se emplea `SMOTE` (*random_state=42*) para equilibrar la distribución de las clases del target.

**Validación y prueba (`preprocessing_eval`)**
- Repite las mismas transformaciones sin volver a ajustar el `DictVectorizer` ni el `StandardScaler`.  
- Aplica *transform()* para garantizar consistencia.  
- Se aseguran las mismas columnas finales que en el conjunto de entrenamiento.


### Diagrama del flujo
Raw Data

├── Limpieza de columnas irrelevantes

├── Mapeo de variables (Country → Continent, Gender, Stress_Level)

├── One-Hot Encoding con DictVectorizer

├── Eliminación de colinealidad (dummy + correlación)

├── Selección de features (Información Mutua)

├── Escalado numérico con StandardScaler

├── Balanceo con SMOTE

└── Datos listos para entrenamiento

## Versionado y Reproducibilidad

- El **preprocesamiento se versiona por separado** del modelo, guardando lo siguiente en MLflow:  
  - `DictVectorizer` (dv.b)  
  - `StandardScaler` (scaler.b)  
  - Features finales (features.pkl)  
- Esto permite reproducir la transformación exacta durante inferencia o reentrenamiento.  
- Todos los pasos usan **semillas fijas (*random_state=42*)** para que sea reproducible entre ejecucuines.

El preprocesamiento genera tres matrices *(X_train_bal, X_val, X_test)* y sus respectivos vectores objetivo *(y_train_bal, y_val, y_test)*, listos para ser usados en los modelos.

# Experimentos de Modelado
---

En esta sección se entrenaron tres modelos distintos con el objetivo de predecir el nivel de estrés a partir de las variables seleccionadas.  

Para cada modelo se utilizó Optuna como motor de optimización de hiperparámetros y MLflow (Databricks) para el seguimiento de experimentos, registro de parámetros, métricas y artefactos.

Se realizaron 10 trials para cada uno de los diferentes modelos.

### Modelo 1: Regresión Logística

El primer modelo entrenado fue una **Regresión Logística**.  
Se optimizaron los siguientes hiperparámetros:

- **Penalty (l1, l2, elasticnet)**: tipo de regularización aplicada al modelo.  
- **C (Regularización)**: controla la fuerza de la regularización (valores pequeños implican regularización más fuerte).  
- **Class weight (None, balanced)**: ajusta los pesos de las clases para manejar desbalance.  
- **Solver (saga)**: algoritmo de optimización utilizado.  
- **Multi class (multinomial)**: especifica el tipo de problema multiclase.

La función  de Optuna entrena el modelo con los hiperparámetros sugeridos y evalúa el rendimiento mediante la métrica *F1* sobre el conjunto de validación.  
El objetivo de la búsqueda en todos los modelos es *maximizar la métrica*.

### Modelo 2: Random Forest

El segundo modelo fue un **Random Forest**.
Los hiperparámetros optimizados fueron:

- **N estimators**: número de árboles en el bosque.  
- **Max depth**: profundidad máxima de los árboles.  
- **Min samples split**: mínimo de muestras necesarias para dividir un nodo.  
- **Min samples leaf**: mínimo de muestras por hoja.  
- **Max features**: número de características consideradas en cada división.  
- **Class weight (None, balanced)**: balanceo de clases.

### Modelo 3: XGBoost

El tercer modelo fue un **XGBoost Classifier**.

Hiperparámetros optimizados:

- **N estimators**: número de árboles en el modelo.  
- **Max depth**: profundidad máxima de los árboles.  
- **Learning rate**: tasa de aprendizaje.  
- **Subsample**: proporción de muestras utilizadas para cada árbol.  
- **Colsample bytree**: proporción de características por árbol.  
- **Gamma**: reducción mínima requerida para hacer una partición.  
- **Min child weight**: peso mínimo de la suma de instancias por nodo.



Para cada uno de los modelos se logearon los experimentos en MLflow, registrando los hiperparámetros, métricas de evaluación tomando F1 como métrica principal y  Accuracy y Log-loss como secundarias y artefactos relevantes (modelos entrenados, preprocesadores).

# Orquestación

## Objetivo

El objetivo de este script es llevar el flujo de trabajo a un entorno operativo, convirtiendo los notebooks en un pipeline más robusto, modular y automatizado. Para esto usamos Prefect como orquestador y MLflow para manejar todo el ciclo de vida del modelo.

El flujo principal orquesta la ejecución secuencial y paralela de las siguientes etapas críticas:

### A. Ingesta y Preparación de Datos
* **Tasks:** `read_csv_task` y `split_task`.
* **Descripción:** Carga los datos crudos y realiza una división estratificada en conjuntos de *Train*, *Validation* y *Test* para asegurar la representatividad de las clases de estrés.

### B. Preprocesamiento Avanzado
* **Tasks:** `preprocess_train_task` y `preprocess_eval_task`.
* **Lógica:** Centraliza la limpieza y transformación de datos, asegurando que el conjunto de validación y prueba se transforme utilizando exactamente los mismos parámetros aprendidos en entrenamiento (evitando *data leakage*).
* **Operaciones Clave:**
    * Ingeniería de variables (Agrupación de Sueño, Mapeo de Continentes).
        - Nota: Es importante mencionar que se decidió agrupar la calidad de sueño en Buena y Mala ya que existía una perfecta correlación entre la calidad del sueño y el nivel de estrés, ocasionando que los modelos no aprendieran sino que repitieran esos patrones y relación directa.
    * Codificación *One-Hot* con `DictVectorizer`.
    * Selección de características mediante **Información Mutua**.
    * Balanceo de clases con **SMOTE** (solo en training).
    * Escalado robusto con `RobustScaler`.

### C. Entrenamiento y Optimización (Tuning)
Se definieron tareas independientes para entrenar y optimizar tres tipos de modelos utilizando **Optuna** para la búsqueda de hiperparámetros:
1.  **Random Forest** (`tune_rf_task`)
2.  **XGBoost** (`tune_xgb_task`)
3.  **Regresión Logística** (`tune_lr_task`)

Cada tarea registra en **MLflow**:
* Métricas de desempeño (F1-Macro, Log Loss, Accuracy).
* Artefactos del modelo y preprocesadores (`dv.b`, `scaler.b`, `features.pkl`).
* La firma del modelo (`infer_signature`) para validar las entradas en producción.

### D. Registro y Promoción de Modelos
* **Task:** `register_models_task`.
* **Automatización:** Al finalizar los entrenamientos, esta tarea consulta MLflow para identificar los mejores experimentos basándose en `f1_macro`.
* **Acción:**
    * Registra automáticamente el mejor modelo como versión **Champion**.
    * Registra el segundo mejor modelo como versión **Challenger**.

- El conjunto de entrenamiento se utilizó para ajustar y entrenar los modelos
- El conjunto de validación se utilizó para seleccionar los mejores hiperparámetros y optimizar
- El conjunto de prueba se utilizó para evaluar el desempeño final de los modelos seleccionados

# Servir el modelo (API FastAPI)

---

## Descripción
El servicio de inferencia se ha implementado utilizando **FastAPI**, exponiendo el modelo de predicción de estrés como un microservicio REST. La separación entre el modelo de ML y la interfaz permite modificar o mejorar cada parte sin afectar a la otra, mejorando la mantenibilidad y la escalabilidad.

El código implementa una **estrategia híbrida** para la gestión de dependencias del modelo, balanceando velocidad y consistencia:

### A. Preprocesadores Estáticos (Locales)
La API carga artefactos de transformación (`dv.b`, `scaler.b`, `features.pkl`) directamente desde el sistema de archivos local (o contenedor).
* **Propósito:** Garantizar la **reproducibilidad exacta** del pipeline de entrenamiento.
* **Componentes:**
    * `DictVectorizer`: Para la codificación *One-Hot* de variables categóricas.
    * `RobustScaler`: Para la normalización de variables numéricas.
    * `features.pkl`: Lista estricta de columnas requeridas para asegurar la alineación con el modelo.

### B. Modelo Dinámico (MLflow Registry)
El modelo predictivo **no se incluye en la imagen Docker**. En su lugar, la API se conecta al registro de modelos de Databricks durante el arranque.
* **Mecanismo:** Utiliza `mlflow.pyfunc.load_model` apuntando a `models:/workspace.default.equipo1-proyecto@Champion`.
* **Ventaja CI/CD:** Permite actualizar el modelo en producción cambiando la etiqueta "Champion" en Databricks sin necesidad de redesplegar el código de la API.

## Validación y Procesamiento 

### Esquema de Entrada (Pydantic)
Se utiliza la clase `InputData` para validar estrictamente los tipos de datos entrantes, rechazando peticiones mal formadas antes de procesarlas.
* **Campos:** `sleep_quality`, `occupation`, `coffee_intake`, `bmi`, etc.

### Pipeline de Preprocesamiento (preprocess)
La función `preprocess` replica las transformaciones realizadas en la etapa de entrenamiento:
1.  **Mapeo de Negocio:**
    * *Países a Continentes:* (e.g., "Mexico" → "America").
    * *Calidad de Sueño:* Agrupación en categorías binarias (e.g., "Poor" → "Bad").
    * *Género:* Mapeo binario (Male: 0, Female: 1).
2.  **Vectorización y Escalado:** Aplica los transformadores cargados (`dv` y `scaler`) y asegura que el DataFrame resultante tenga exactamente las columnas que el modelo espera, rellenando valores faltantes con 0 si es necesario.

## Flujo interno

Como se mencionó anteriormente, la API fue implementada con FastAPI y expone un servicio de predicción basado en el modelo registrado como Champion en el MLflow Model Registry. Al iniciar, el servicio realiza la carga del modelo, así como de los artefactos de preprocesamiento (dv, scaler y features) necesarios para garantizar que los datos enviados por el usuario se transformen exactamente de la misma forma que durante el entrenamiento.

El endpoint utiliza Pydantic para validar el input recibido, lo cual asegura que todos los campos requeridos tengan el tipo correcto y que la estructura del request sea consistente. Una vez validada la entrada, la API ejecuta un preprocesamiento consistente: mapeo de categorías, ingeniería de variables, transformación con DictVectorizer, alineación con el conjunto de features finales y escalado numérico. Esto permite mantener compatibilidad total entre los datos de producción y el pipeline utilizado durante el entrenamiento del modelo.

## Manejo de errores

El servicio implementa manejo de errores explícito, devolviendo códigos HTTP adecuados:

- 200 (OK): la predicción se genera correctamente.

- 500 (Internal Server Error): problemas al cargar preprocesadores o el modelo, errores inesperados en la transformación o predicción.

- Errores específicos se propagan mediante HTTPException, mostrando mensajes claros.

Además, como se dijo anteriormente en los endpoints, se incluye un endpoint /health que permite verificar si el modelo y los preprocesadores fueron cargados correctamente, facilitando monitoreo y despliegue.

## Ejemplo de request/response

**Request:**

```json
{
  "sleep_quality": "Good",
  "occupation": "Student",
  "coffee_intake": 2.0,
  "physical_activity_hours": 1.5,
  "country": "Mexico",
  "bmi": 23.4,
  "alcohol_consumption": "Yes",
  "age": 25,
  "gender": "Female",
  "heart_rate": 75,
  "smoking": "No"
}
```
**Response:**

```json
{
  "prediction": 1,
}
```


Esto iniciará un servidor local y abrirá automáticamente la aplicación en el navegador, permitiendo al usuario interactuar con el formulario y consultar predicciones directamente a través de la API

El código completo de la aplicación se encuentra en el archivo `main.py`, dentro de la carpeta `src/frontend`.

---

# Contenerización y Orquestación (Docker & Docker Compose)
---

## Estrategia 
Para garantizar la portabilidad del sistema, se empaquetaron tanto el Backend (API) como el Frontend (UI) en contenedores Docker ligeros y aislados.

### Imágenes Base y Optimización
Ambos servicios utilizan python:3.13-slim como imagen base.
* **¿Por qué Slim?** Reduce drásticamente el tamaño final de la imagen al excluir librerías innecesarias, mejorando la velocidad de despliegue y reduciendo la superficie de ataque de seguridad.
* **Configuración Regional:** Se configura explícitamente la zona horaria a America/Mexico_City dentro del contenedor para asegurar coherencia en los logs y registros temporales.

### Definición de Contenedores

#### A. Contenedor de API (Backend)
El contenedor de la API se construye con una imagen base ligera de Python 3.13-slim. Se copian los archivos necesarios, se instalan las dependencias desde `requirements.txt`, y se expone el puerto 8000 para servir la API con Uvicorn.

* **Dependencias:** Instala las librerías listadas en `requirements.txt`.

    Dentro de las dependencias mínimas se incluyen:
    - fastapi==0.121.0
    - mlflow==3.4.0
    - mlflow_skinny==3.4.0
    - mlflow_tracing==3.4.0
    - pandas==2.3.3
    - pydantic==2.12.4
    - python-dotenv==1.2.1
    - scikit-learn==1.7.2
        
* **Artefactos:** Copia la carpeta `preprocessor/` (que contiene dv.b, scaler.b, etc.) dentro de la imagen. Esto hace que el contenedor sea autosuficiente para realizar inferencias.
* **Ejecución:** Lanza el servidor uvicorn en el puerto 8000.

Para ver el detalle y el código completo del Dockerfile, se puede encontrar en `src/backend/Dockerfile`, junto con el archivo `requirements.txt` que contiene las dependencias.

#### B. Contenedor de UI (Frontend)
El contenedor de la UI también utiliza Python 3.13-slim como base. Copia el script `main.py`, instala Streamlit y Requests, y expone el puerto 8501 para la interfaz web.
* **Dependencias:** 

    Las dependecias mínimas incluyen:
    - Requests==2.32.5
    - streamlit==1.51.0

Para ver el detalle y el código completo del Dockerfile, se puede encontrar en `src/frontend/Dockerfile`, junto con el archivo `requirements.txt` que contiene las dependencias.

### Docker Compose

Se utiliza `docker-compose.yaml` para gestionar el ciclo de vida de la aplicación multi-contenedor, permitiendo levantar todo el ecosistema con un solo comando.

### Configuración del Servicio (docker-compose.yaml)

1.  **Servicio API (api):**
    * Construye la imagen desde `./src/backend`.
    * Carga automática de variables de entorno desde el archivo .env, necesarias para las credenciales de Databricks/MLflow.
    * Expone el puerto 8000.

2.  **Servicio UI (ui):**
    * Construye la imagen desde `./src/frontend`.
    * Expone el puerto 8501.

### Configuración de Red (network_mode: host)
Se optó por utilizar el modo de red host.
* **Justificación:** Esto permite que los contenedores compartan la pila de red del host local. Facilita que la UI (Streamlit) se comunique con la API simplemente llamando a `http://localhost:8000` o 127.0.0.1, sin necesidad de configurar redes Docker adicionales o enlaces entre contenedores.

El archivo `docker-compose.yaml` se encuentra en la raíz del proyecto.

---

# Despliegue en la nube (AWS o HuggingFace Spaces)
---

## Estrategia 
Para la puesta en producción se seleccionó Hugging Face Spaces, una plataforma que permite desplegar aplicaciones de Machine Learning de forma rápida y sencilla, además de ofrecer soporte nativo para contenedores Docker. Se optó por una arquitectura basada en microservicios desacoplados, manteniendo la API separada de la interfaz de usuario (UI).

Se crearon y configuraron dos *Spaces* independientes utilizando el SDK de Docker:

### A. Backend: API de Inferencia (`proyecto-final`)
Este espacio aloja el servidor FastAPI y es responsable de procesar las solicitudes y ejecutar el modelo.
* **Proceso de Configuración:**
    1. Creación del repositorio del Space.
    2. Clonado local del repositorio.
    3. Carga de archivos esenciales:
        * `api.py`: Código fuente de la API.
        * `Dockerfile`: Instrucciones para construir la imagen (Python 3.13-slim).
        * `requirements.txt`: Dependencias del backend.
        * `preprocessor/`: Carpeta con los artefactos (`dv.b`, `scaler.b`).
    4. **Configuración de Puertos:** Se editó el `README.md` (metadatos YAML) para exponer el puerto `8000`, permitiendo el tráfico entrante hacia uvicorn.

### B. Frontend: Interfaz de Usuario (`proyecto-final-ui`)
Este espacio aloja la aplicación de Streamlit y actúa como cliente del Backend.
* **Proceso de Configuración:**
    1. Creación del repositorio del Space.
    2. Clonado local del repositorio.
    3. Carga de archivos esenciales:
        * `ui.py`: Código fuente de la interfaz gráfica (Streamlit).
        * `Dockerfile`: Instrucciones para construir la imagen de UI.
        * `requirements.txt`: Dependencias (streamlit, requests).
    4. **Configuración de Puertos:** Se editó el `README.md` para exponer el puerto `8501` (puerto por defecto de Streamlit).

## Integración
La comunicación entre ambos servicios se realiza vía HTTP. La Interfaz de Usuario (`proyecto-final-ui`) envía las peticiones de predicción a la URL pública generada por el espacio del API (`proyecto-final`), garantizando que ambos componentes puedan escalar o actualizarse de forma independiente en la infraestructura de Hugging Face.

`Link:` https://huggingface.co/spaces/sarahbltn/proyecto-final-ui 

## Evidencia del servicio corriendo

![Hugging Face](docs\screenshots\HuggingFace1.jpeg)

![Hugging Face](docs\screenshots\HuggingFace2.jpeg)

# Conclusiones
---

Durante la exploración y modelado de los datos, identificamos patrones críticos que definieron el rumbo del proyecto:

* **Redundancia en variables:** Descubrimos que la variable Caffeine_mg tenía una correlación extremadamente alta con Coffee_Intake. Aprendimos que mantener ambas introducía multicolinealidad sin agregar valor predictivo, por lo que eliminarla fue clave para simplificar el modelo.
* **Calidad de los datos sintéticos:** La variable Health_Issues presentaba más del 50% de valores nulos. Esto nos enseñó la importancia de auditar la calidad del dato antes de modelar; imputar tantos datos habría introducido demasiado ruido, por lo que la eliminación fue la decisión correcta.
* **Límite de desempeño (Plateau):** Tanto el modelo *Champion* como el *Challenger*  convergieron en un F1-Score similar de **0.69**. lo que indica que ambos están explotando al máximo la información disponible en este dataset sintético. Esto sugiere que incrementar la complejidad del modelo no aporta mejoras debido a la limitada señal presente en los datos.

Más allá de los números, el desarrollo del pipeline nos dejó aprendizajes técnicos importantes:

1.  **El valor de la Orquestación:** Implementar **Prefect** transformó el flujo de trabajo. Pasar de ejecutar notebooks manuales a tener tareas modulares (read_csv, preprocess, tune) nos permitió iterar más rápido y con menos errores humanos.
2.  **Prevención de *Data Leakage*:** Aprendimos a separar estrictamente el preprocesamiento de entrenamiento del de validación. Guardar el DictVectorizer y el RobustScaler como artefactos en MLflow garantizó que los nuevos datos se transformen exactamente igual que los de entrenamiento.
3.  **La importancia de la Portabilidad:** La contenerización con **Docker** resolvió los problemas de compatibilidad. Al empaquetar la API y el Frontend por separado, logramos que el despliegue en Hugging Face Spaces fuera fluido, independientemente del entorno local de desarrollo.

Para llevar este proyecto al siguiente nivel en una segunda iteración, proponemos:

* **Incorporar Datos Reales:** Como el proyecto se desarrolló con datos sintéticos, es probable que las relaciones entre el consumo de café y el nivel de estrés estén demasiado simplificadas. Contar con datos biométricos reales ayudaría a capturar patrones más precisos y mejorar el F1-Score.
* **Ingeniería de Características más compleja:** Podríamos explorar interacciones entre variables, como *Cafeína por Índice de Masa Corporal (BMI)*, para ver si el efecto del café varía según la complexión física.

# Referencias
---

Caffeine Intake and Mental Health
Macritchie, K. (2020). https://kamac-sheff.github.io/Caffeine/Caffeine.html 

Grosso, G., Godos,J., Galvano,F.
and  Giovannucci, E. Coffee, Caffeine, and Health Outcomes: An Umbrella Review. https://yseulthb.github.io/pdf/grosso2017.pdf 

Global Coffee Health Dataset. (2025, August 22). https://www.kaggle.com/datasets/uom190346a/global-coffee-health-dataset/data
