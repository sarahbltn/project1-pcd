# Introducción
---

El consumo de café es un producto que puede generar placer, energía, incluso ayudar al estado de animo, sin embargo este también puede afectar la salud, incluyendo el riesgo de enfermedades cardiovasculares, metabólicas y neurológicas e incluso, en ciertos casos, generar dependencia.

El problema general que aborda este proyecto es entender la relación entre el consumo de café y diferentes factores de salud como el nivel de estrés, lo cual es relevante ya que ayuda a prevenir enfermedades, creando decisiones informadas dentro de la industria de nutrición.

El alcance de esta primera entrega se centra en realizar un análisis exploratorio descriptivo (EDA) para identificar patrones generales y entender los datos con los que se estarán trabajando, también se realizará una organización de los datos donde se hará la limpieza de los mismos.

Se trabajará con el dataset *GlobalCoffeeHealth* desde Kaggle el cual contiene 10.000 registros sintéticos que reflejan patrones reales de consumo de café, hábitos de sueño y resultados de salud en 20 países como Alemania, Brasil, México, entre otros. 

Fue obtenido del siguiente link: https://www.kaggle.com/datasets/uom190346a/global-coffee-health-dataset/data

# Planteamiento del Problema
---

**a) Formulación del problema**

El problema principal consiste en determinar cómo el consumo de café y otros hábitos de vida influyen en los niveles de estrés de las personas en diferentes países. 

Pregunta:

¿Se podrá predecir los niveles de estrés a partir de variables relacionadas con consumo de café, hábitos de salud y estilo de vida?

**b) Variables clave y supuestos iniciales**

Entradas -> Coffee Intake, Sleep Quality, Age, Country, BMI, Heart Rate, Alcohol Consumption, Smoking Status, Exercise Frequency

Salidas -> Stress Level

* El consumo de cafeína tiene un efecto directo o indirecto sobre el estrés.
* Existen diferencias significativas entre países o continentes que podrían explicar variaciones culturales o de estilo de vida.

**c) Impacto esperado**

Comprender la relación entre el consumo de café, la calidad del sueño y los niveles de estrés puede ayudar a informar decisiones de salud pública y recomendaciones dietéticas. Además, identificar patrones globales puede contribuir a estrategias personalizadas para mejorar el bienestar.

**d) Riesgos conocidos**

Se pueden observar correlaciones bajas entre las variables, lo que dificultaría la creación de modelos predictivos efectivos.


# EDA
---

**Qué se exploró y por qué**

Se realizó una exploración inicial para:

1. Identificar valores faltantes y evaluar el dataset.

2. Revisar tipos de variables (numéricas y categóricas) para definir estrategias de análisis posteriores.

3. Analizar distribuciones, sesgos, rangos y outliers en las variables numéricas, para ver valores atípicos.

4. Calcular correlaciones entre variables para ver relaciones significativas.

5. Evaluar la presencia de duplicados entre más.

**Hallazgos principales**

- Se cuenta con 10,000 registros y 16 columnas, entre ellas 9 son numéricas y 6 categóricas (sin contar el ID).
- Se puede observar que la columna *Health Issue* tiene muchos valores faltantes (más del 50%) lo que se tomarán decisiones al respecto.
- No se encontraron duplicados.
- Variables con más outliers: *Heart_Rate (50)*, *BMI (40)*, *Coffee_Intake (30)* y *Caffeine_mg (20)*.
- Se observa una alta correlación entre *Coffee_Intake* y *Caffeine_mg*.
