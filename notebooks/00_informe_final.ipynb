# Introducción
---

El consumo de café es un producto que puede generar placer, energía, incluso ayudar al estado de ánimo, sin embargo este también puede afectar la salud, incluyendo el riesgo de enfermedades cardiovasculares, metabólicas y neurológicas e incluso, en ciertos casos, generar dependencia.

El problema general que aborda este proyecto es entender la relación entre el consumo de café y diferentes factores de salud sobretodo con el nivel de estrés, lo cual es relevante ya que ayuda a prevenir enfermedades, creando decisiones informadas dentro de la industria de nutrición.

El alcance de esta primera entrega se centra en realizar un análisis exploratorio descriptivo (EDA) para identificar patrones generales y entender los datos con los que se estarán trabajando, también se realizará una organización de los datos donde se hará la limpieza de los mismos para poder trabajar con datos limpios.

Se trabajará con el dataset *GlobalCoffeeHealth* desde Kaggle el cual contiene 10.000 registros sintéticos que reflejan patrones reales de consumo de café, hábitos de sueño, niveles de estrés y resultados de salud en 20 países como Alemania, Brasil, México, entre otros. 

El dataset fue obtenido del siguiente link: https://www.kaggle.com/datasets/uom190346a/global-coffee-health-dataset/data

# Preprocesamiento de Datos
---

El preprocesamiento se realiza por modulos mediante dos funciones principales:  
`preprocessing_train()`, para entrenamiento y `preprocessing_eval()`, para validación y prueba.  
Ambas transforman los datos de forma consistente, solo la primera ajusta los objetos (**DictVectorizer**, **StandardScaler**, **SMOTE**) y define las features finales.

## Flujo del Prepocesamiento
**Entrenamiento (`preprocessing_train`)**
1. **Limpieza de columnas:**  
   Se eliminan las variables: *Health_Issues*, *Caffeine_mg*, *Sleep_Hours*, *Sleep_Quality*
2. **Filtrado de datos:**  
   Se excluyen registros con *Gender = "Other"*.
3. **Codificación categórica manual:**  
   - *Country* se transforma a *Continent* mediante un mapeo (América, Europa, Asia, Oceanía).  
   - *Gender* se convierte a binaria (*0 = Male*, *1 = Female*).  
   - *Stress_Level* se codifica ordinalmente (*Low=0*, *Medium=1*, *High=2*).
4. **Codificación categórica general:**  
   Se aplica un `DictVectorizer` para convertir todas las categorías en variables binarias.
5. **Reducción de colinealidad:**  
   - Se elimina un dummy por categoría (para evitar colinealidad perfecta).  
   - Se eliminan variables con alta correlación (|r| > 0.9) usando Pearson.
6. **Selección de características:**  
   Se calculó la **información mutua** (*mutual_info_classif*) para identificar las variables con mayor dependencia con el target.  
   Se conservaron las **20 más informativas**, de las cuales *15* pasaron los filtros de correlación y redundancia.
7. **Estandarización:**  
   Se aplica `StandardScaler` a las variables seleccionadas.
8. **Balanceo de clases:**  
   Se emplea `SMOTE` (*random_state=42*) para equilibrar la distribución de las clases del target.

**Validación y prueba (`preprocessing_eval`)**
- Repite las mismas transformaciones sin volver a ajustar el `DictVectorizer` ni el `StandardScaler`.  
- Aplica *transform()* para garantizar consistencia.  
- Se aseguran las mismas columnas finales que en el conjunto de entrenamiento.


### Diagrama del flujo
Raw Data

├── Limpieza de columnas irrelevantes

├── Mapeo de variables (Country → Continent, Gender, Stress_Level)

├── One-Hot Encoding con DictVectorizer

├── Eliminación de colinealidad (dummy + correlación)

├── Selección de features (Información Mutua)

├── Escalado numérico con StandardScaler

├── Balanceo con SMOTE

└── Datos listos para entrenamiento

## Versionado y Reproducibilidad

- El **preprocesamiento se versiona por separado** del modelo, guardando lo siguiente en MLflow:  
  - `DictVectorizer` (dv.pkl)  
  - `StandardScaler` (scaler.pkl)  
  - Lista de features finales (features.json)  
- Esto permite reproducir la transformación exacta durante inferencia o reentrenamiento.  
- Todos los pasos usan **semillas fijas (*random_state=42*)** para que sea reproducible entre ejecucuines.

El preprocesamiento genera tres matrices *(X_train_bal, X_val, X_test)* y sus respectivos vectores objetivo *(y_train_bal, y_val, y_test)*, listos para ser usados en los modelos.

# Experimentos de Modelado
---

En esta sección se entrenaron tres modelos distintos con el objetivo de predecir el nivel de estrés a partir de las variables seleccionadas.  

Para cada modelo se utilizó Optuna como motor de optimización de hiperparámetros y MLflow (Databricks) para el seguimiento de experimentos, registro de parámetros, métricas y artefactos.

Se realizaron 10 trials para cada uno de los diferentes modelos.

### Modelo 1: Regresión Logística

El primer modelo entrenado fue una **Regresión Logística**.  
Se optimizaron los siguientes hiperparámetros:

- **Penalty (l1, l2, elasticnet)**: tipo de regularización aplicada al modelo.  
- **C (Regularización)**: controla la fuerza de la regularización (valores pequeños implican regularización más fuerte).  
- **Class weight (None, balanced)**: ajusta los pesos de las clases para manejar desbalance.  
- **Solver (saga)**: algoritmo de optimización utilizado.  
- **Multi class (multinomial)**: especifica el tipo de problema multiclase.

La función  de Optuna entrena el modelo con los hiperparámetros sugeridos y evalúa el rendimiento mediante la métrica *F1* sobre el conjunto de validación.  
El objetivo de la búsqueda en todos los modelos es *maximizar la métrica*.

### Modelo 2: Random Forest

El segundo modelo fue un **Random Forest**.
Los hiperparámetros optimizados fueron:

- **N estimators**: número de árboles en el bosque.  
- **Max depth**: profundidad máxima de los árboles.  
- **Min samples split**: mínimo de muestras necesarias para dividir un nodo.  
- **Min samples leaf**: mínimo de muestras por hoja.  
- **Max features**: número de características consideradas en cada división.  
- **Class weight (None, balanced)**: balanceo de clases.

### Modelo 3: XGBoost

El tercer modelo fue un **XGBoost Classifier**.

Hiperparámetros optimizados:

- **N estimators**: número de árboles en el modelo.  
- **Max depth**: profundidad máxima de los árboles.  
- **Learning rate**: tasa de aprendizaje.  
- **Subsample**: proporción de muestras utilizadas para cada árbol.  
- **Colsample bytree**: proporción de características por árbol.  
- **Gamma**: reducción mínima requerida para hacer una partición.  
- **Min child weight**: peso mínimo de la suma de instancias por nodo.
